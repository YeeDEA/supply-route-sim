```python
import tkinter as tk
from tkinter import messagebox
from matplotlib import pyplot as plt
import numpy as np
import random
import folium
import os
from PIL import Image

class Supply:
    def __init__(self, name, quantity, effect_on_stats):
        self.name = name
        self.quantity = quantity
        self.effect_on_stats = effect_on_stats

    def consume(self, amount):
        if amount <= self.quantity:
            self.quantity -= amount
            return True
        return False


class Soldier:
    def __init__(self, name, hunger, health, stress, hygiene):
        self.name = name
        self.hunger = hunger
        self.health = health
        self.stress = stress
        self.hygiene = hygiene

    def consume_supply(self, supply):
        if supply.name == "food" and self.hunger < 100:
            self.hunger += supply.effect_on_stats["hunger"]
            supply.consume(1)


class SimulationApp:
    def __init__(self, master):
        self.master = master
        self.master.title("Military Logistics Simulation")

        self.canvas = tk.Canvas(master, width=800, height=600, bg="lightblue")
        self.canvas.pack()

        self.buildings = []
        self.threats = []
        self.map_center = [37.5665, 126.978]  # Default center (Seoul)
        self.create_ui()

    def create_ui(self):
        # Add buttons for actions
        self.btn_add_building = tk.Button(self.master, text="Add Building", command=self.add_building)
        self.btn_add_building.pack(side=tk.LEFT)

        self.btn_add_threat = tk.Button(self.master, text="Add Threat", command=self.add_threat)
        self.btn_add_threat.pack(side=tk.LEFT)

        self.btn_capture = tk.Button(self.master, text="Capture Screenshot", command=self.capture_screenshot)
        self.btn_capture.pack(side=tk.LEFT)

        self.btn_compare_paths = tk.Button(self.master, text="Compare Paths", command=self.compare_paths)
        self.btn_compare_paths.pack(side=tk.LEFT)

        self.btn_run_simulation = tk.Button(self.master, text="Run Simulation", command=self.run_simulation)
        self.btn_run_simulation.pack(side=tk.LEFT)

    def add_building(self):
        x, y = random.uniform(37.564, 37.570), random.uniform(126.973, 126.980)  # Random coordinates
        self.buildings.append((x, y))
        # Optionally, draw the building on the canvas
        self.canvas.create_rectangle(100, 100, 140, 140, fill="green")

    def add_threat(self):
        x, y = random.uniform(37.564, 37.570), random.uniform(126.973, 126.980)  # Random coordinates
        self.threats.append((x, y))
        # Optionally, draw the threat on the canvas
        self.canvas.create_oval(150, 150, 170, 170, fill="red")

    def capture_screenshot(self):
        # Create a Folium map and add buildings and threats
        folium_map = folium.Map(location=self.map_center, zoom_start=15)

        for building in self.buildings:
            folium.Marker(location=building, popup='Building', icon=folium.Icon(color='blue')).add_to(folium_map)

        for threat in self.threats:
            folium.Marker(location=threat, popup='Threat', icon=folium.Icon(color='red')).add_to(folium_map)

        filename = "military_simulation_map.html"
        folium_map.save(filename)
        os.startfile(filename)  # Open in browser
        messagebox.showinfo("Map", f"Map saved as {filename}")

    def compare_paths(self):
        # Simulate path comparison (simplified)
        plt.figure(figsize=(10, 5))
        plt.plot(np.random.rand(10), np.random.rand(10), 'g-', label='Initial Path')
        plt.plot(np.random.rand(10), np.random.rand(10), 'r-', label='Altered Path')
        plt.title('Path Comparison')
        plt.legend()
        plt.show()

    def run_simulation(self):
        # Simulate the supply and soldier logic
        soldiers = [Soldier(f'Soldier {i}', random.randint(70, 100), 100, 0, 100) for i in range(5)]
        supply = Supply("food", 100, {"hunger": 10})

        for soldier in soldiers:
            soldier.consume_supply(supply)

        # Display soldier states
        for soldier in soldiers:
            print(f"{soldier.name} - Hunger: {soldier.hunger}")

        # Simulate supply distribution over a week
        days = ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5", "Day 6", "Day 7"]
        supply_distribution = [random.randint(10, 20) for _ in days]

        # Plotting supply distribution
        plt.figure(figsize=(10, 5))
        plt.bar(days, supply_distribution, color='blue')
        plt.title('Weekly Supply Distribution')
        plt.xlabel('Days')
        plt.ylabel('Supply Amount')
        plt.show()


if __name__ == "__main__":
    root = tk.Tk()
    app = SimulationApp(root)
    root.mainloop()
```
